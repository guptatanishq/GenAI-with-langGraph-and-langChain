```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int , b:int):
    """
        it will add the two numbers 
        Args:
            a : Number one
            b : Number twp


    """
    return a+b

@tool
def sub_number(a:int , b:int):
    """
        it will subtract the two numbers 
        Args:
            a : Number one
            b : Number twp


    """
    return a-b
```

```python
add_number.invoke({"a": 12, "b": 3})
sub_number.invoke({"a": 12, "b": 3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model = 'gemini-3.5-flash-lite')
```

```python
agent = create_agent(
    model = llm,
    tools = [add_number, sub_number],
    system_prompt = "you are a math teacher and uses tool for calculation"
    
)
```

```python
res = agent.invoke({"messages": [{"role":"user", "content": "what is 2+3"}]})
print(res)
```

```python
for res in res['messages']:
    print(res)
    print('\n')
```
