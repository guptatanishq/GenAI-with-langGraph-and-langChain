```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm =  ChatGoogleGenerativeAI(model = 'gemini-3.5-flash-lite')
```

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description = "Complete name")
    email: str = Field(description = "Email address")
    age : int  = Field(description = "Age")

structure_llm = llm.with_structured_output(ResponseStructure)
```

```python
text = "Hello my name is tanishq ,"\
    "email is gupta.tanishq2004@gmail.com and age is 22"
```

```python
res = structure_llm.invoke(f"please give me my name, age and email address from this text: {text}" )
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title: str = Field(description="movie title")
    year: int = Field(description="movie year")


class AllMovies(BaseModel):
    movies: List[Movies]


structure_llm = llm.with_structured_output(AllMovies)
structure_llm.invoke("give me the name of spider man movies")
```
